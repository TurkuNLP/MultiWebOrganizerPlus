```python
import json
```

```python
path_to_data = "/scratch/project_465002530/users/tarkkaot/LLM_document_descriptors/data/weborganizer/topic_format_edu.jsonl"

with open(path_to_data, "r") as f:
    data = []
    for line in f:
        doc = json.loads(line)
        data.append({
            "text": doc["document"],
            "doc_id": doc["doc_id"]
        })
```

```python
with open("../data/FineWebSample.jsonl", "w") as f:
    for line in data:
        f.write(json.dumps(line, ensure_ascii=False))
        f.write("\n")
```
